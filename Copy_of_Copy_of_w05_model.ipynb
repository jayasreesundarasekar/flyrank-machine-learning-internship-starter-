## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Split design:
# Use a time-aware split so earlier observations are used for training
# and later observations are used for validation.
#
# This avoids using future information to predict the past and better
# represents how the model would be used in practice.

df = df.sort_values("date").copy()

split_date = df["date"].quantile(0.8)

train = df[df["date"] < split_date].copy()
test = df[df["date"] >= split_date].copy()

print("Train rows:", len(train))
print("Validation rows:", len(test))
print("Train date range:", train["date"].min(), "to", train["date"].max())
print("Validation date range:", test["date"].min(), "to", test["date"].max())
```

## 3. Train + compare vs my baseline

*Same data, same metric, same split as your Week-4 baseline. Show the table.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# IMPORTANT:
# Replace these with the same legitimate features used by your Week-4 baseline.
FEATURES = [
    "feature_1",
    "feature_2",
    "feature_3",
    "feature_4",
    "feature_5"
]

TARGET = "label"

X_train = train[FEATURES]
y_train = train[TARGET]

X_test = test[FEATURES]
y_test = test[TARGET]

# Train the model.
model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000))
])

model.fit(X_train, y_train)

# Generate predictions.
pred = model.predict(X_test)

# Evaluate using the same metric as the Week-4 baseline.
model_f1 = f1_score(y_test, pred, zero_division=0)

print("Logistic Regression F1:", model_f1)

# Replace this with the actual Week-4 baseline metric.
baseline_f1 = YOUR_WEEK4_BASELINE_F1

comparison = pd.DataFrame({
    "method": [
        "Week-4 baseline",
        "Logistic Regression"
    ],
    "f1": [
        baseline_f1,
        model_f1
    ]
})

comparison
```

## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Identify incorrect predictions.
errors = test.loc[y_test.index].copy()
errors["actual"] = y_test
errors["predicted"] = pred

errors = errors[errors["actual"] != errors["predicted"]]

print("Total validation rows:", len(test))
print("Incorrect predictions:", len(errors))
print("Error rate:", len(errors) / len(test))

# Inspect the errors.
errors[FEATURES + ["actual", "predicted"]].head(20)

# Inspect Logistic Regression coefficients.
# Larger absolute coefficients indicate features the model relies on more strongly.

coefficients = pd.DataFrame({
    "feature": FEATURES,
    "coefficient": model.named_steps["model"].coef_[0]
})

coefficients["absolute_weight"] = coefficients["coefficient"].abs()

coefficients.sort_values(
    "absolute_weight",
    ascending=False
)
```
