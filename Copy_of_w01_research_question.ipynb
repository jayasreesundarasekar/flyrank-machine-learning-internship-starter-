## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
elif os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")

import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

# Number 1: scale + base rate
print(df.shape[0], "pages | declining rate:", round(df["is_declining_label"].mean(), 3))

# Number 2: how many pages my Week 2 hand rule actually covers, and its hit rate
stale_visible = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]
print("Stale+visible pages:", len(stale_visible), f"({len(stale_visible)/len(df):.1%} of all pages)")
print("Declining rate within that group:", round(stale_visible["is_declining_label"].mean(), 3))

# Number 3: how skewed impressions are -- why getting the top of the queue right matters most
cutoff = df["impressions_90d"].quantile(0.90)
top_share = df.loc[df["impressions_90d"] >= cutoff, "impressions_90d"].sum() / df["impressions_90d"].sum()
print(f"Top 10% of pages hold {top_share:.1%} of all impressions")
```
