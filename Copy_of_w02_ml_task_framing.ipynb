## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Task type: Scoring / Ranking.
#
# The output is NOT a single yes/no per page -- it's a ranked queue: every page
# gets a continuous opportunity score, and editors work down the list from the
# top. Under the hood this can be *implemented* with a binary classifier (like
# the starter pipeline's decision tree/random forest), but the classifier's
# probability output is used as a score to rank pages, not as a hard label.
# That distinction matters because editors have fixed weekly capacity -- they
# need the BEST 50 pages, not just a flag on all 30,000.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Primary metric: Precision@50.
#
# Why this one and not overall accuracy or ROC AUC: the action this supports is
# an editor working through a fixed-size weekly queue, not a global classification
# decision. What matters is "of the top 50 pages the model surfaces, how many are
# actually worth an editor's time" -- that's exactly what Precision@50 measures,
# and it's the metric Week 2 already used to compare my hand rule against the tree.
# ROC AUC / average precision are useful secondary checks (they show whether the
# model has real signal across the whole ranking, not just the top), but Precision@50
# is the number that maps directly onto the decision this lane supports.
```
